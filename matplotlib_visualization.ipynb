# Iris dataset analysis and visualization

## 1. Importing required libraries

```python
import pandas as pd
import matplotlib.pyplot as plt                                                                                                #visualization library
from sklearn.datasets import load_iris
```

## 2. Loading the Iris dataset

```python
iris = load_iris(as_frame=True)
df= iris.frame
```

```python
df
```

## 3. Basic line plot

```python
#simply plotting a line graph
#simple introduction to plotting before you start plotting actual iris data
x = [1,2,3]                                                                    
y = [2,4,6]  
plt.plot(x,y) 
plt.show()       # to display the line graph
```

## 4. Line plot using Dataset Features

```python
#line graph
plt.plot(
    df["sepal width (cm)"]
)
plt.xlabel("Observation")                                                                                                    # naming x axis
plt.ylabel("Sepal width (cm)")                                                                                                     # naming y axis
plt.title("Sepal Width across Observations")                                                                                                         # naming graph
plt.show()
```

## 5. Bar chart of Flower Type Distribution

```python
count = df["target"].value_counts()

plt.bar(

 count.index,
 count.values
 )

plt.xlabel("Types")
plt.ylabel("No of Flowers")

plt.title("Bar plot")

plt.show()
```

## 6. Scatter Plot of Petal Measurements

```python
#scatter plotting : for grouping identification/ similarity sharing
plt.scatter(                                                     
    df["petal length (cm)"],
    df["petal width (cm)"]
)
plt.xlabel("Petal length (cm)")
plt.ylabel("Petal width (cm)")
plt.title("Petal length vs Petal width")
plt.show()
```

## 7. Histogram: distribution of petal length

```python
#histogram: to check distribution
plt.hist(
     df["petal length (cm)"],
    bins=10
)
plt.xlabel("petal length")
plt.ylabel("frequency")
plt.title("histogram plot")
plt.show()

    
```

## 8. Combining multiple visualizations

## 9. Customizing the Bar Chart

```python
count = df["target"].value_counts()

plt.figure(figsize=(5,5))

plt.bar(

 count.index,
 count.values,
 color=("#800080","#008080","#000080")
    
 )

plt.xlabel("Types", fontsize=12)
plt.ylabel("No of Flowers", fontsize=12)

plt.title("Bar plot",
         fontsize=12,
         fontweight='bold')

plt.show()
```

## 10. Pie chart of Flower type Distribution

```python
count = df["target"].value_counts()

plt.figure(figsize=(10,5))

plt.pie(
    count.values,
    labels=count.index,
    autopct="%1.1f%%" ,
    colors=["#800080","#008080","#000080"]
)
plt.title("pie plot",
         fontsize=12,
         fontweight='bold')

plt.show()
    
```

## 11. Conclusion

The Iris dataset was explored using Pandas and visualized using Matplotlib. Different visualization techniques, including line plots, bar charts, scatter plots, histograms, and pie charts, were used to understand the dataset from different perspectives.

These visualizations help identify patterns, distributions, and relationships between different features of the dataset.
