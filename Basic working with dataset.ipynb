# Flower Dataset Analysis

## 1. Importing required libraries

```python
import pandas as pd
```

```python
import numpy as np
```

## 2. Loading the Dataset

The dataset is loaded from a CSV file using Pandas. The loaded data is stored in a DataFrame for further analysis.

```python
df=pd.read_csv("flower_pandas_numpy_dataset.csv")
df
```

## 3. Selecting Relevant Features

Specific columns are selected from the dataset to focus the analysis on sepal length and petal length.

```python
df[
 [
 "sepal length (cm)",
 "petal length (cm)"
 ]
 ]
```

## 4. Analyzing Target Distribution

The frequency of each target class is calculated to understand how the observations are distributed among the different flower types.

```python
df["target"].value_counts()
```

## 5. Group-wise Feature Analysis

The average petal length is calculated for each target class using group-wise aggregation.

```python
df.groupby("target")["petal length (cm)"].mean()
```

## 6. Sorting the Dataset

The dataset is sorted according to petal length in descending order to identify observations with the largest petal measurements.

```python
df.sort_values(
 "petal length (cm)", 
 ascending= False
 ).head(5)
```

## 7. Renaming the Target Column

The target column is renamed to `Types` to make its purpose more descriptive and readable.

```python
df=df.rename(columns={"target":"Types"})
```

```python
df.columns
```

## 8. Converting Target Values into Meaningful Labels

The numerical target values are mapped to descriptive flower type labels to make the dataset easier to interpret.

```python
Types_map = {
    0: "Pink",
    1: "White",
    2: "Blue"
}

df["Types_map"] = df["Types"].map(Types_map)
df
```

## 9. Conclusion

In this notebook, the flower dataset was loaded and explored using Pandas and NumPy. Different data manipulation techniques were demonstrated, including column selection, value counting, grouping, sorting, renaming columns, and mapping numerical values to meaningful labels.

These operations provide a basic foundation for exploring and understanding datasets before applying machine learning techniques.
